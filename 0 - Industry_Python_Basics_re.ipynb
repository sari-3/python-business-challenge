## Exercise 1 (BASIC): a calculator for young learners

BrightStart, an education company, has asked you to build a simple calculator that helps children practise the four main operations: addition, subtraction, multiplication and division.

A child enters a list of numbers and one of four operation names: `"add"`, `"subtract"`, `"multiply"` or `"divide"`. The calculator applies the operation from left to right, starting with the first number. For example:

* `simple_math([60, 20, 15], "subtract")` calculates 60 − 20 − 15 and returns `25`
* `simple_math([120, 2, 15], "divide")` calculates 120 ÷ 2 ÷ 15 and returns `4.0`

Write a function `simple_math(numbers, operation)`.

You can assume that:

* the list always contains at least one number (if it contains only one, return that number);
* the operation is always one of the four names above; and
* the calculation never divides by zero.

> **INPUT:** a list of numbers and a string

> **OUTPUT:** integer or float

```python
def simple_math(numbers, operation):
    answer = numbers[0]

    for num in numbers[1:]:
        if operation == "add":
            answer = answer + num
        elif operation == "subtract":
            answer = answer - num
        elif operation == "multiply":
            answer = answer * num
        else:
            answer = answer / num
    return answer
```

## Exercise 2 (BASIC): tracking event replies

You are collecting replies from colleagues for an upcoming team-building workshop. Each colleague submits their preferred first name and one of three decisions:

| Decision | Meaning |
|---|---|
| `"T"` | Tentative |
| `"A"` | Accept |
| `"D"` | Decline |

Write a function `attend_tracker(name, decision)`.

**Step 1: check the name.** A name is correctly formatted if it starts with a capital letter and every other letter is lowercase, for example `Melinda`. If the name is not correctly formatted, return `"Did you mean <corrected name>?"`, whatever the decision. The corrected name starts with a capital letter, with every other letter lowercase. For example, `attend_tracker("AlEx", "T")` returns `"Did you mean Alex?"`.

**Step 2: reply to the decision.** If the name is correctly formatted, return the message for the decision:

| Decision | Return |
|---|---|
| `"T"` | `"We hope to see you there, <name>!"` |
| `"A"` | `"Can't wait, <name>!"` |
| `"D"` | `"Hope you can make it next time, <name>!"` |

You can assume the decision is always `"T"`, `"A"` or `"D"`.

> **Note: names with a hyphen, an apostrophe or a space.** Names such as `Anne-Marie`, `O'Brien` or `Mary Jane` are not covered by the checks, and there is more than one acceptable way to handle them. Decide how your function should treat them and add a comment to your code explaining your choice. Your reviewer will review how you have handled them.

> **INPUT:** 2 strings

> **OUTPUT:** string

```python
def attend_tracker(name, decision):
    # step 1 - check name is capitalised, using string formatting
    if name != name.capitalize():
        return f"Did you mean {name.capitalize()}?"
    # step 2 - reply to decision
    if decision == "T":
        return f"We hope to see you there, {name}!"
    elif decision == "A":
        return f"Can't wait, {name}!"
    else:
        return f"Hope you can make it next time, {name}!"
```

## Exercise 3 (BASIC): rating product names

Your company's communications director wants a catchy name for its newest product. The team has suggested many names, and you have been asked to score them. Rather than score each one by hand, you will write a function to do it.

Every name starts with a score of **80**. The four rules below then adjust it, and a character can score under more than one rule.

| Rule | What to check | Score change |
|---|---|---|
| 1. Uppercase letters | Count the uppercase letters. If there are 3 or fewer, each one scores. If there are more than 3, none of them score. | +5 for each uppercase letter |
| 2. Length | A name of 4 to 10 characters (inclusive) is ideal. Outside that range, each character too many or too few loses points. | −3 for each character outside the range |
| 3. Special characters | A special character is any character that is not a letter or a digit, including a space. `!` and `?` are allowed and do not count. | −5 for each special character |
| 4. Marketable letters | The letters X, Y, Z, Q and V, in uppercase or lowercase | +2 for each marketable letter |

**Length examples.** A 13-character name is 3 characters too long, so it loses 9 points. A 2-character name is 2 characters too short, so it loses 6 points.

**Worked example.** `"AquaQuark"` scores 94:

* Rule 1: 2 uppercase letters (A and Q): +10
* Rule 2: 9 characters, inside the range: no change
* Rule 3: no special characters: no change
* Rule 4: 2 marketable letters (q and Q): +4

Write a function `name_rater(name)` that returns the score.

> **INPUT:** string

> **OUTPUT:** integer

```python
def name_rater(name):
    # starts with 80
    rating = 80
    points_deduct = 0
    # apply the rules
    #rule 1
    if sum(1 for c in name if c.isupper()) <= 3:
        rating = rating + (5 * sum(1 for c in name if c.isupper()))
    else:
        rating = rating

    # rule 2
    if len(name) >10:
        points_deduct = (len(name) - 10) *3
        rating = rating - points_deduct
    elif len(name) <4:
        points_deduct = (4 - len(name)) *3
        rating = rating - points_deduct
    else:
        rating = rating

    # rule 3
    for c in name:
        if not c.isalnum() and c not in "!?":
            rating = rating - 5

    # rule 4
    for c in name:
        if c.upper() in "XYZQV":
            rating = rating + 2
            
    return rating
```
