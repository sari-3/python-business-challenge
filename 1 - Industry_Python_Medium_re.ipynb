## Exercise 1 (MEDIUM): calculating take-home pay

The country of Oongland has introduced a new banded income tax system. Your family has asked you to work out how much the four of you will take home this year, after tax.

Tax is charged **per band**: each part of an income is taxed at the rate for the band it falls in. Moving into a higher band never reduces anyone's take-home pay, because only the part of the income above a band boundary is taxed at the higher rate.

| Band | Part of the income | Tax rate |
|---|---|---|
| 1 | Up to and including 15,000 | 0% |
| 2 | Above 15,000, up to and including 22,000 | 10% |
| 3 | Above 22,000, up to and including 36,500 | 25% |
| 4 | Above 36,500 | 40% |

All amounts are in Oonglian pounds.

**Worked example.** An income of 25,000 pays:

* 0% on the first 15,000: 0
* 10% on the next 7,000 (from 15,000 to 22,000): 700
* 25% on the remaining 3,000 (from 22,000 to 25,000): 750

The total tax is 1,450, so the take-home pay is 23,550.

Write a function `oong_calc(incomes)` that takes a list of the four family members' yearly incomes and returns their **total take-home pay**, rounded to 2 decimal places with `round(total, 2)`.

Return `"Invalid input"` if:

* the input is not a list;
* the list does not contain exactly 4 items; or
* any item is not a number (an integer or a decimal).

You can assume every number is 0 or more.

> **INPUT:** list

> **OUTPUT:** float (or the string `"Invalid input"`)

```python
def oong_calc(incomes):
    # first validate the pay inputs

    # if the input is not a list
    if not isinstance(incomes, list):
        return "Invalid input"
    # if the list doesnt have 4 items
    if len(incomes) != 4:
        return "Invalid input"
    # if an item is not a numeric
    for i in incomes:
        if not isinstance(i, (int, float)) or isinstance(i, bool):
            return "Invalid input"

    # calculate the tax
    thp = 0
    for i in incomes:
        tax = 0

        #band 1: 0% tax
        if i <= 15000:
            tax = 0
        #band 2+ 10% tax for 15000-22000
        elif i <=22000:
            tax = (i - 15000) * 0.10
        #band 3: 25% tax for 22000-36500 so the 10% of band 2 + for band 3
        elif i <=36500:
            tax = (7000 * 0.10) + ((i - 22000) * 0.25)
        #band 4: 40% tax for above 36500 so the 10% of band 2 + the 25% of band 3 + for band 4
        else:
            tax = (7000 * 0.10) + (14500 * 0.25) + ((i - 36500) * 0.40)

        #work out the take home pay
        thp = thp + (i - tax)

    return round(thp,2)
```

## Exercise 2 (MEDIUM): a fun fact for every phone number

To improve customer satisfaction, a mobile network wants to greet each new customer with a fun fact about their phone number. It sells two types of number:

| Type | Format |
|---|---|
| Business | 6 digits, starting with `0` |
| Classic | 11 digits, starting with `07` (the UK mobile format) |

Write a function `num_funfact(number)` that takes a phone number as a string.

If the input is not a valid number, return `"Invalid number!"`. A valid number is a string made only of the digits `0` to `9`, with no spaces or other characters, that matches one of the two formats above.

For a valid number, return a fun fact followed by the number's type, in the form `"<fun fact> You have a <type> number."`, where `<type>` is `business` or `classic`. Use the **first** fun fact in this table that applies:

| Order | The fun fact applies when | Fun fact |
|---|---|---|
| 1 | The number contains every digit from 0 to 9 | `Wow, your number contains all 10 digits!` |
| 2 | One digit appears at least 3 times anywhere in the number | `Your number contains the digit <digit> at least 3 times!` |
| 3 | The number contains a run of at least 3 consecutive digits in ascending order | `You have a consecutive streak in your number ;)` |
| 4 | None of the above | `Your number is great!` |

**Notes**

* For fun fact 2, the repeated digits do not need to be next to each other. If more than one digit appears at least 3 times, report the one that appears first in the number.
* For fun fact 3, each digit in the run must be exactly one more than the digit before it, for example `678` or `3456`. Descending runs such as `876` do not count, and `0` does not follow `9`, so `890` does not count.

For example, `num_funfact("090345")` returns `"You have a consecutive streak in your number ;) You have a business number."`

> **INPUT:** string

> **OUTPUT:** string

```python
def num_funfact(number):
    # first check if input is a string
    if not isinstance(number, str):
        return "Invalid number!"

    # next check input is only digits 0-9
    if not number.isdigit():
        return "Invalid number!"

    # now checking the number type and that the length of te string is 6 or 11(business or classic)
    if len(number) == 6 and number.startswith("0"):
        number_type = "business"
    elif len(number) == 11 and number.startswith("07"):
        number_type = "classic"
    else:
        return "Invalid number!"

    # outputting the fun facts
    # fact 1 - containing all digits 0-9 
    if all(str(i) in number for i in range(10)):
        fact = "Wow, your number contains all 10 digits!"
    
    #fact 2 - repeated digit anywhere in the str
    elif any(number.count(i) >=3 for i in number):
        for i in number:
            if number.count(i) >=3:
                fact = f"Your number contains the digit {i} at least 3 times!"
                # so only the first repeated number is outputted
                break

    #fact 3 - 3 consec nums in asc order
    elif any(int(number[i+1]) == int(number[i]) + 1
            and int(number[i+2]) == int(number[i+1]) +1
            for i in range(len(number) - 2)):
        fact = "You have a consecutive streak in your number ;)"

    # rule 4 - none of the above
    else:
        fact = "Your number is great!"

    return f"{fact} You have a {number_type} number."
    
```

## Exercise 3 (MEDIUM): generating usernames

You are loading customer records into a data warehouse, and each customer needs a username. The source system sends each customer's **full name** as a tuple `(first name, surname)` and their **birth year** as an integer from 1911 to 2010 inclusive. Assume every customer has exactly one first name and one surname.

Build the username in four steps:

1. The first letter of the first name, in uppercase.
2. An underscore.
3. The surname, in lowercase.
4. The last 2 digits of the birth year, keeping any leading zero (2007 gives `07`).

For example, `gen_user([("Haskell", "Covus"), 1982])` returns `"H_covus82"`, and `gen_user([("Franz", "Manzo"), 1958])` returns `"F_manzo58"`.

**Complication 1: duplicate usernames.** Two customers can produce the same username: Ana Smith and Anastasia Smith, both born in 1990, would both be `A_smith90`. When a username already exists, the source system sends a third item, an integer from 0 to 99 inclusive. Add another underscore and that number to the end of the username. For example, `gen_user([("Anastasia", "Smith"), 1990, 27])` returns `"A_smith90_27"`.

You do not need to check for duplicates yourself: the extra number is only sent when it is needed.

**Complication 2: shuffled input.** The source system does not always send the items in the same order. All three of these inputs return `"A_smith90_27"`:

* `[("Anastasia", "Smith"), 1990, 27]`
* `[1990, 27, ("Anastasia", "Smith")]`
* `[27, ("Anastasia", "Smith"), 1990]`

The input is always a list containing one tuple and one or two integers. You do not need to handle invalid input.

> **INPUT:** list of 2 or 3 items

> **OUTPUT:** string

```python
def gen_user(customer_details):
    # checking the inputs and where the name and bday is
    # but need a defult dup bc some inputs might not be dups
    dup = None
    for item in customer_details:
        if isinstance(item, tuple):
            first_name = item[0]
            surname = item[1]
        elif item >=1911:
            birth_year = item
        else:
            dup = item

    # generating the username
    username = f"{first_name[0].upper()}_{surname.lower()}{str(birth_year)[-2:]}"

    # now need to add the num if it is a dup
    if dup is not None:
        username = f"{username}_{dup}"

    return username
```
